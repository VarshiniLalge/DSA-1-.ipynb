```python
class Solution:
    def twoSum(self, nums, target):
        seen = {}

        for i in range(len(nums)):
            complement = target - nums[i]

            if complement in seen:
                return [seen[complement], i]

            seen[nums[i]] = i
```

Step-by-step

1. seen = {}

Creates an empty dictionary to store:

number → index

For example:

{2: 0, 7: 1}

means 2 is at index 0 and 7 is at index 1.

2. Loop through the numbers

for i in range(len(nums)):

i represents the current index.

3. Find the required number

complement = target - nums[i]

This calculates what number is needed to reach the target.

Example:

target = 9
current number = 2

complement = 9 - 2 = 7

So we need 7.

4. Check if we already saw it

if complement in seen:

If 7 is already in the dictionary, we found the two numbers.

return [seen[complement], i]

This returns their indexes.

5. Store the current number

seen[nums[i]] = i

If the required number isn't found, store the current number and its index for later.

Example
nums = [2, 7, 11, 15]
target = 9
2 → needs 7 → not found → store 2
7 → needs 2 → found!

So:

return [0, 1]
🧠 Remember

Current number + Complement = Target

Complement = Target - Current number

The dictionary helps us find the complement quickly, giving the solution O(n) time complexity.
